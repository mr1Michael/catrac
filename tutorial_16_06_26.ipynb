```python
# bubble sort

list_to_sort = []
number_of_items = len(list_to_sort)

# start with the first item
# compare to the next one
# carry the greater of the 2 to compare to the next one

i = 0

while i < number_of_items - 1:
    if list_to_sort[i] > list_to_sort[i+1]:
        # swap them
        temp = list_to_sort[i]
        list_to_sort[i] = list_to_sort[i+1]
        list_to_sort[i+1] = temp
```
